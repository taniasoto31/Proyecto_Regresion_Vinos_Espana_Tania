# 📊 Informe de Rendimiento y Evaluación del Modelo Predictivo

**Proyecto:** Predicción de Precios de Vino Español (*Spanish Wine Quality*)   
**Fecha:** Noviembre 2026  
**Modelo Evaluado:** Random Forest Regressor (Optimizado vía GridSearchCV)

---

## 1. Resumen Ejecutivo

El presente informe detalla la evaluación cuantitativa y cualitativa del modelo de *Machine Learning* desarrollado para predecir el precio ($€$) de los vinos en el mercado español. A partir del análisis exploratorio y del ajuste de hiperparámetros, se ha obtenido un modelo basado en **Random Forest Regressor** estructurado para equilibrar capacidad predictiva y generalización.

---

## 2. Métricas de Rendimiento de Regresión

Para evaluar la capacidad predictiva del modelo y verificar la presencia o ausencia de *overfitting* (sobreajuste), se comparan las métricas clave en los conjuntos de **Entrenamiento (Train)** y **Prueba (Test)**.

### 2.1 Tabulación de Métricas

| Métrica | Conjunto Train | Conjunto Test | Brecha / Variación |
| :--- | :---: | :---: | :---: |
| **$R^2$ (Coeficiente de Determinación)** | 0.5235 | 0.5059 | $3.36\%$ |
| **RMSE (Root Mean Squared Error)** | $112.20\text{ €}$ | $114.46\text{ €}$ | $+2.01\%$ |
| **MAE (Mean Absolute Error)** | $27.28\text{ €}$ | $33.22\text{ €}$ | $+21.75\%$ |

### 2.2 Diagnóstico de Sobreajuste (Overfitting)

* **Análisis del $R^2$:** La diferencia relativa de rendimiento entre el conjunto de entrenamiento ($R^2 = 0.5235$) y el conjunto de prueba ($R^2 = 0.5059$) es del **$3.36\%$**.
* **Conclusión:** Al ser la brecha inferior al **$5\%$**, el modelo demuestra una sólida capacidad de generalización con datos no vistos, descartando problemas severos de memoria o sobreajuste (*overfitting*).
* **Matiz sobre el MAE:** El error absoluto medio aumenta un **$21.75\%$** entre Train y Test ($27.28\text{ €}$ → $33.22\text{ €}$). El $R^2$ y el RMSE están dominados por los vinos de precio elevado, por lo que conviene interpretarlos junto con el MAE, que refleja mejor el error típico por botella.

## 3. Importancia de las Características (*Feature Importance*)

El análisis de la relevancia de variables generada por el ensemble de árboles (agrupando las variables categóricas codificadas) arroja la siguiente distribución de pesos descriptivos:

```
Puntuación (Rating)     [########################################] 62.4%
Num. Reseñas (Reviews)  [===============] 23.5%
Región (Region)         [=====] 7.1%
Tipo de Vino (Type)     [====] 5.4%
Cuerpo (Body)           [=] 1.5%
Acidez (Acidity)        [=] 0.1%
```

### Principales Hallazgos:
1. **Puntuación (`rating`):** Es la variable de mayor peso intrínseco ($62.4\%$). La percepción de calidad de los evaluadores/usuarios es la principal variable explicativa del precio final, aunque debe interpretarse como relación predictiva y no causal.
2. **Volumen de Opiniones (`num_reviews`):** Representa un $23.5\%$ de la importancia, reflejando que la popularidad y el reconocimiento del mercado influyen positivamente en la valoración económica.
3. **Región y Tipo de Vino (`region` y `type`):** Aportan en conjunto un $12.5\%$ ($7.1\%$ y $5.4\%$ respectivamente), capturando diferencias de origen y estilo que influyen en el posicionamiento de precio.
4. **Atributos Organolépticos (`body` y `acidity`):** El cuerpo ($1.5\%$) y la acidez ($0.1\%$) tienen una aportación marginal en este modelo, por lo que no son determinantes para diferenciar el precio.

## 4. Análisis de Residuos e Interpretación Gráfica

### 4.1 Residuos ($\text{Real} - \text{Predicción}$)
El análisis de los errores residuales por tramo de precio permite identificar comportamientos atípicos en las predicciones:

* **Segmento Estándar (≤ $150\text{ €}$):** Concentra el $93.1\%$ de los vinos de prueba ($1{,}180$ de $1{,}267$), con un MAE de $17.52\text{ €}$ y un residuo medio de $-12.83\text{ €}$ (ligera sobreestimación). Los residuos se agrupan alrededor del valor $0$, con estimaciones razonablemente estables en el segmento comercial estándar, aunque el error típico sigue siendo relevante frente al precio mediano ( 29.15 € ) y la dispersión crece con el precio predicho (heterocedasticidad).

* **Segmento Intermedio ($150\text{ €}$ – $500\text{ €}$):** Con solo $62$ vinos, el MAE asciende a $140.27\text{ €}$. El residuo medio ($+11.87\text{ €}$) es cercano a $0$ únicamente porque errores de signo opuesto se compensan; el modelo es poco fiable en este tramo.
* **Sesgo en Vinos de Alta Gama (*Outliers*):** En botellas de colección o edición limitada que superan los $500\text{ €}$ ($25$ vinos), el modelo **subestima el precio** de forma sistemática (residuo medio de $+492.61\text{ €}$ y MAE de $508.89\text{ €}$). Esto es una consecuencia esperada debido a la baja frecuencia de datos en valores extremos y al efecto de compresión de varianza típico de los métodos basados en árboles de decisión.

### 4.2 Comportamiento Predicción vs. Valores Reales
Al representar la gráfica de **Valores Reales vs. Valores Predichos** frente a la línea ideal de $45^\circ$ ($y = x$):
* Se observa una alineación consistente a lo largo de la diagonal en la mayor densidad de datos (precios bajos).
* La dispersión aumenta progresivamente a medida que el precio real se incrementa: las predicciones se saturan en torno a los $900\text{ €}$ aunque el precio real alcance los $1{,}750\text{ €}$. También se observan casos puntuales de vinos de bajo precio con predicciones elevadas.

## 5. Recomendaciones Técnicas e Itinerario Futuro

1. **Transformación Logarítmica:** Aplicar $\log(y)$ sobre el precio para estabilizar la varianza de los valores extremos (*outliers*), evaluando su efecto por tramo de precio antes de adoptarla..
2. **Ingeniería de Características Adicionales:** Explorar la interacción entre la bodega (`winery`) y la denominación de origen (`region`) para capturar la reputación de marca.
